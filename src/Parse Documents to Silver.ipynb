Use [ai_parse_document function](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/ai_parse_document) to extract document layouts as VARIANT output and review the results

```sql
SELECT
  path,
  ai_parse_document(content)
FROM READ_FILES(:sourceResume, format => 'binaryFile');
```

Take the Statement above and write it to a Silver Delta Table

```sql
INSERT INTO IDENTIFIER(:silvertable)
SELECT
  path,
  ai_parse_document(content) AS content,
  current_timestamp() AS modifiedon
FROM READ_FILES(:sourceResume, format => 'binaryFile');
```

Move PDF files to a processed subfolder

```sql
%python

resumefiles = dbutils.widgets.get("sourceResume")

files = dbutils.fs.ls(resumefiles)
for file in files:
    if file.name.endswith(".pdf"):
        dbutils.fs.mv(file.path, f"{resumefiles}/processed/{file.name}", True)
```
