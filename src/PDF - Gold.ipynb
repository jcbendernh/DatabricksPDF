The following example uses [ai_query](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/ai_query) to separate each top-level field and review the results

```sql
SELECT
  path,
  content,
  ai_query(
    'databricks-meta-llama-3-3-70b-instruct',
    concat('Extract just the name from the following text. Do not inlcude any other text', content),
    returnType => 'STRING'
  ) AS name,
  CAST(
    ai_query(
      'databricks-meta-llama-3-3-70b-instruct',
      concat('Extract just the job experience from the following text. Do not inlcude any other text', content),
      returnType => 'STRING'
    ) AS VARIANT
  ) AS experience,
  CAST(
    ai_query(
      'databricks-meta-llama-3-3-70b-instruct',
      concat('Extract just the education experience from the following text. Do not inlcude any other text', content),
      returnType => 'STRING'
    ) AS VARIANT
  ) AS education,
  CAST(
    ai_query(
      'databricks-meta-llama-3-3-70b-instruct',
      concat('Extract just the certification experience from the following text. Do not inlcude any other text', content),
      returnType => 'STRING'
    ) AS VARIANT
  ) AS certification
FROM IDENTIFIER(:silvertable)
```

Create the Gold table if it does not exist.

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:goldtable) (
  path STRING,
  content STRING,
  name STRING,
  experience VARIANT,
  education VARIANT,
  certification VARIANT,
  modifiedon TIMESTAMP
);
```

Insert new records into the gold table

```sql
  INSERT INTO IDENTIFIER(:goldtable)
  
  SELECT
    path,
    content,
    ai_query(
      'databricks-meta-llama-3-3-70b-instruct',
      concat('Extract just the name from the following text. Do not inlcude any other text', content),
      returnType => 'STRING'
    ) AS name,
    CAST(
      ai_query(
        'databricks-meta-llama-3-3-70b-instruct',
        concat('Extract just the job experience from the following text. Do not inlcude any other text', content),
        returnType => 'STRING'
      ) AS VARIANT
    ) AS experience,
    CAST(
      ai_query(
        'databricks-meta-llama-3-3-70b-instruct',
        concat('Extract just the education experience from the following text. Do not inlcude any other text', content),
        returnType => 'STRING'
      ) AS VARIANT
    ) AS education,
    CAST(
      ai_query(
        'databricks-meta-llama-3-3-70b-instruct',
        concat('Extract just the certification experience from the following text. Do not inlcude any other text', content),
        returnType => 'STRING'
      ) AS VARIANT
    ) AS certification,
      current_timestamp() AS modifiedon
  FROM IDENTIFIER(:silvertable)
```
